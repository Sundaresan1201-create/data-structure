```python
data={
    'Rollno':[101,102,103,104,105,106,10,108,109,110],
    'Name':['Arun','Priya','Rahul','anantha','nitin','sundar','mani','agash','arivu','nithish'],
    'Marks':[85,90,78,80,80,80,80,80,80,80]}
df=pd.DataFrame(data)
print(df)
print(df.columns)
print(df.index)
print(df.dtypes)
```

```python
df=pd.DataFrame(data,index=['A','B','C','A','A','A','A','A','A','A'])
print(df)
```

```python
df['Grade']=['A','A+','C','A','A','A','A','A','A','A']
print(df[['Name','Marks']])
```

```python
print(df.iloc[0])
print(df.iloc[[0,2]])
print(df.iloc[1])
print(df.iloc[[0,2]])
```

```python
df['Marks']=df['Marks']+5
print(df)
```
